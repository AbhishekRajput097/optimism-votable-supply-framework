```python
import pandas as pd

# Read the CSV file
file_name = "../../../data/votable_supply_framework_data/avpi_data_collection_agora/proposals_weightage_data.csv"
df = pd.read_csv(file_name)

# List of columns to process
columns_to_divide = [
    "Sum of FOR weightage",
    "Sum of AGAINST weightage",
    "Sum of ABSTAIN weightage",
    "Total weightage"
]

# Apply the division to the specified columns
for column in columns_to_divide:
    df[column] = df[column].astype(float) / (10**18)

# Save the updated DataFrame to a new CSV file
output_file = "../../../data/votable_supply_framework_data/avpi_data_collection_agora/proposals_weightage_data_processed.csv"
df.to_csv(output_file, index=False)

print(f"Processed data saved to {output_file}")
```

```python
import pandas as pd

# File names
file1 = "../../../data/votable_supply_framework_data/avpi_data_collection_agora/proposals_weightage_data_processed.csv"
file2 = "../../../data/votable_supply_framework_data/avpi_data_collection_agora/proposals_data_original.csv"
output_file = "../../../data/votable_supply_framework_data/avpi_data_collection_agora/merged_proposals_data.csv"

# Read the first and second CSV files
df1 = pd.read_csv(file1)
df2 = pd.read_csv(file2)

# Rename the 'id' column in the second file to match 'proposalId' for merging
df2.rename(columns={"id": "proposalId"}, inplace=True)

# Merge the DataFrames on 'proposalId' and include only the 'endTime' column from df2
merged_df = pd.merge(df1, df2[["proposalId", "endTime"]], on="proposalId", how="left")

# Save the merged DataFrame to a new CSV file
merged_df.to_csv(output_file, index=False)

print(f"Merged data saved to {output_file}")
```

```python
import pandas as pd

# File name
file_name = "../../../data/votable_supply_framework_data/avpi_data_collection_agora/merged_proposals_data.csv"
output_file = "merged_proposals_data_cleaned.csv"

# Read the CSV file
df = pd.read_csv(file_name)

# Convert the 'endTime' column to date format (YYYY-MM-DD)
df['endTime'] = pd.to_datetime(df['endTime']).dt.date

# Save the updated DataFrame to a new CSV file
df.to_csv(output_file, index=False)

print(f"Cleaned data saved to {output_file}")
```
