```python
import pandas as pd
from datetime import datetime, timedelta

# Load the data from the CSV file
data = pd.read_csv("../../../../data/votable_supply_framework_data/avpi_data_collection_agora/proposals_data/pdv.csv", parse_dates=["Date"])
data.sort_values(by="Date", inplace=True)

# Prepare a new DataFrame to hold the daily data
start_date = data["Date"].min()
end_date = data["Date"].max()
date_range = pd.date_range(start=start_date, end=end_date, freq='D')

# Create a DataFrame with all dates and fill missing VP values with forward fill
daily_data = pd.DataFrame({"Date": date_range})
daily_data = daily_data.merge(data, on="Date", how="left")
daily_data["VP"] = daily_data["VP"].fillna(method='ffill')

# Save the result to a new CSV file
daily_data.to_csv("PDV_daily.csv", index=False)

print("Daily data has been written to 'PDV_daily.csv'")
```
