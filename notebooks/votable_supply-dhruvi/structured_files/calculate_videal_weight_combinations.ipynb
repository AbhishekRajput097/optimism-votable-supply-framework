```python
import os
import glob
import pandas as pd
```

```python
# Define file paths and parameters
file_paths = glob.glob("../data/weight_combinations_results/*.csv")
parameters_file = "../../../data/votable_supply-dhruvi/data/indexes.csv"
output_folder = "../data/weight_combinations_results_with_videal"

os.makedirs(output_folder, exist_ok=True)

parameters_df = pd.read_csv(parameters_file)

if "Date" not in parameters_df.columns or "Votable Supply" not in parameters_df.columns:
    raise ValueError("Parameters_Values.csv must contain 'Date' and 'Votable Supply' columns.")

parameters_df = parameters_df[["Date", "Votable Supply"]]

alpha = 0.15

for file_path in file_paths:
    df = pd.read_csv(file_path)
    
    if "Date" in df.columns and "F" in df.columns:
        merged_df = pd.merge(df, parameters_df, on="Date", how="left")
        
        if merged_df["Votable Supply"].isnull().any():
            print(f"Warning: Missing 'Votable Supply' values after merging for file {file_path}.")
        
        merged_df.rename(columns={"Votable Supply": "Voriginal"}, inplace=True)
        merged_df["Videal"] = round(merged_df["Voriginal"] * (1 + alpha * (merged_df["F"] - 1)), 2)
        
        # Save the processed file
        file_name = os.path.basename(file_path)
        output_path = os.path.join(output_folder, file_name)
        merged_df.to_csv(output_path, index=False)
        
        print(f"Processed and saved: {file_name}")
    else:
        print(f"Skipping {file_path} - Required columns not found.")

print("All files processed successfully!")
```
