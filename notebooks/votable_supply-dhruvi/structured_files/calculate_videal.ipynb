```python
import pandas as pd
```

### Calculate Videal

```python
# Function to find the optimal alpha value from the file
def find_alpha(alpha_file):
    alpha_data = pd.read_csv(alpha_file)
    max_row = alpha_data.loc[alpha_data["Percentage Within Range"].idxmax()]
    
    # Extract the alpha value for the highest percentage
    corresponding_alpha = max_row["Alpha"]
    return corresponding_alpha

# Main function
if __name__ == "__main__":
    alpha_file = "../../../data/votable_supply-dhruvi/data/optimal_alpha_ideal_weights.csv"  # Input file path
    data_file = "../../../data/votable_supply-dhruvi/data/adjustment_factor_results.csv"  # Input data file path
    output_path = "../../../data/votable_supply-dhruvi/data/videal_results.csv"  # Output file path
    
    # Get the alpha value from the file
    alpha = find_alpha(alpha_file)
    print(f"Alpha value used for Videal calculation: {alpha}")
    
    # Read the data
    data = pd.read_csv(data_file)
    
    # Calculate Videal
    data['Videal'] = data['Votable Supply'] * (1 + alpha * (data['F'] - 1))
    
    # Save the results to a CSV file
    data.to_csv(output_path, index=False)
    print(f"Results saved successfully to {output_path}")
```

### Calculate Monthly Videal

```python
data = pd.read_csv("../../../data/votable_supply-dhruvi/data/videal_results.csv")

data["Date"] = pd.to_datetime(data["Date"])

data["Month_Year"] = data["Date"].dt.to_period("M")

monthly_videal = data.groupby("Month_Year")["Videal"].mean().reset_index()

monthly_videal["Month_Year"] = monthly_videal["Month_Year"].dt.to_timestamp()

monthly_videal.rename(columns={"Videal": "Monthly Average Videal"}, inplace=True)

monthly_videal.to_csv("../../../data/votable_supply-dhruvi/data/monthly_videal.csv", index=False)

print("Monthly average Videal calculated and saved to 'monthly_videal.csv'")
```
