```python
import pandas as pd
```

### Calculate Actual Voting Power Index (AVPI)

```python
avpi_raw_data = pd.read_csv("../../../data/votable_supply-dhruvi/data/avpi_raw_data.csv")
```

```python
avpi_raw_data.info()
```

```python
avpi_raw_data['AVPI'] = round(avpi_raw_data['Actual Votable Supply '] / avpi_raw_data['30 Day Avg Total Volume (USD)'], 6)
```

```python
avpi_raw_data.to_csv("../../../data/votable_supply-dhruvi/data/avpi.csv", index=False)
```

### Calculate Liquidity Activity Ratio (LAR)

```python
lar_raw_data = pd.read_csv("../../../data/votable_supply-dhruvi/data/lar_raw_data.csv")
```

```python
lar_raw_data.info()
```

```python
lar_raw_data['LAR'] = lar_raw_data['Total Volume (USD)'] / lar_raw_data['Circulating Supply']
```

```python
lar_raw_data.to_csv("../../../data/votable_supply-dhruvi/data/lar.csv", index=False)
```

### Calculate Participation Ratio (PR)

```python
pr_raw_data = pd.read_csv("../../../data/votable_supply-dhruvi/data/pr_raw_data.csv")
```

```python
pr_raw_data.info()
```

```python
pr_raw_data['PR'] = round(pr_raw_data['Votable Supply'] / pr_raw_data['Circulating Supply'], 6)
```

```python
pr_raw_data.to_csv("../../../data/votable_supply-dhruvi/data/pr.csv", index=False)
```

### Calculate Price Stability Index (PSI)

```python
psi_raw_data = pd.read_csv("../../../data/votable_supply-dhruvi/data/lar_raw_data.csv")
```

```python
psi_raw_data.info()
```

```python
psi_raw_data["7-day Avg Price"] = psi_raw_data["Price (USD)"].rolling(window=7, min_periods=1).mean()
```

```python
psi_raw_data["PSI"] = psi_raw_data["7-day Avg Price"] / psi_raw_data["Price (USD)"]
```

```python
psi_raw_data.to_csv("../../../data/votable_supply-dhruvi/data/psi.csv", index=False)
```

### Calculate Voting Power Index (VPI)

```python
vpi_raw_data = pd.read_csv("../../../data/votable_supply-dhruvi/data/vpi_raw_data.csv")
```

```python
vpi_raw_data.info()
```

```python
vpi_raw_data['VPI'] = vpi_raw_data['Votable Supply'] / vpi_raw_data['30 Day Avg Total Volume (USD)']
```

```python
vpi_raw_data.to_csv("../../../data/votable_supply-dhruvi/data/vpi.csv", index=False)
```
