```python
import requests
import csv
from datetime import datetime

# API endpoint
url = "https://api.llama.fi/protocol/silo-finance"

# Fetch data from the API
response = requests.get(url)
if response.status_code == 200:
    data = response.json()
    tokens_data = data.get("tokens", [])
else:
    print("Failed to fetch data from the API.")
    exit()

# Prepare CSV file
csv_file = "../../../data/votable_supply_framework/lending_protocols_op/silo-finance_data.csv"
fields = ["Date", "OP"]

# Open CSV file for writing
with open(csv_file, mode="w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerow(fields)  # Write header

    # Process each token data entry
    for entry in tokens_data:
        timestamp = entry.get("date")
        token_values = entry.get("tokens", {})
        op_value = token_values.get("OP", 0)

        # Convert timestamp to human-readable format
        readable_date = datetime.utcfromtimestamp(timestamp).strftime('%Y-%m-%d %H:%M:%S')

        # Write row to CSV
        writer.writerow([readable_date, op_value])

print(f"Data has been successfully written to {csv_file}.")
```
