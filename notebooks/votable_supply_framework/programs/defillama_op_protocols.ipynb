```python
import requests
import csv

# API endpoint
api_url = "https://api.llama.fi/protocols"

# Fetch data from API
response = requests.get(api_url)

if response.status_code == 200:
    data = response.json()

    # Filter data for protocols with "Optimism" in "chains"
    filtered_data = [
        {
            "id": protocol.get("id"),
            "name": protocol.get("name"),
            "url": protocol.get("url"),
            "description": protocol.get("description"),
            "chain": protocol.get("chain"),
            "twitter": protocol.get("twitter"),
            "methodology": protocol.get("methodology"),
            "tvl": protocol.get("tvl"),
            "slug": protocol.get("slug"),
        }
        for protocol in data
        if "Optimism" in protocol.get("chains", [])
    ]

    # Sort data by TVL in descending order
    sorted_data = sorted(filtered_data, key=lambda x: x["tvl"], reverse=True)

    # Write data to CSV file
    csv_file = "../../../data/votable_supply_framework/programs/optimism_protocols.csv"
    csv_headers = ["id", "name", "url", "description", "chain", "twitter", "methodology", "tvl", "slug"]

    with open(csv_file, mode="w", newline="", encoding="utf-8") as file:
        writer = csv.DictWriter(file, fieldnames=csv_headers)
        writer.writeheader()
        writer.writerows(sorted_data)

    print(f"Data successfully written to {csv_file}")
else:
    print(f"Failed to fetch data. Status code: {response.status_code}")
```

```python
import requests
import json

# Define the API endpoint
api_url = "https://api.llama.fi/protocol/aave-v3"

try:
    # Make a GET request to the API endpoint
    response = requests.get(api_url)
    response.raise_for_status()  # Raise an error if the request was unsuccessful

    # Parse the JSON response
    data = response.json()

    # Define a recursive function to explore JSON objects and sub-objects
    def explore_json(json_obj, indent=0):
        if isinstance(json_obj, dict):
            for key, value in json_obj.items():
                print(" " * indent + f"{key}: {type(value).__name__}")
                explore_json(value, indent + 4)
        elif isinstance(json_obj, list):
            print(" " * indent + f"List of {len(json_obj)} items")
            if json_obj:  # Check if the list is not empty
                explore_json(json_obj[0], indent + 4)

    # Explore the JSON data structure
    print("Exploring JSON data structure:")
    explore_json(data)

except requests.exceptions.RequestException as e:
    print(f"An error occurred while making the API request: {e}")
except json.JSONDecodeError as e:
    print(f"An error occurred while parsing the JSON response: {e}")
```
