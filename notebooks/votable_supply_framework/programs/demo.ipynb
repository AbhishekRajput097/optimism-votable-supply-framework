```python
import requests
import csv
from datetime import datetime

# Read the slugs from the CSV file
with open('../../../data/votable_supply_framework/programs/optimism_protocols.csv', mode='r', newline='', encoding='utf-8') as slug_file:
    slug_reader = csv.DictReader(slug_file)  # Use DictReader to access columns by name
    slugs = [row['slug'] for row in slug_reader]  # Extract the slug from the 'slug' column

# Loop through each slug and fetch the data
for slug in slugs:
    api_url = f"https://api.llama.fi/protocol/{slug}"

    try:
        # Fetch data from the API
        response = requests.get(api_url)
        response.raise_for_status()  # Ensure the request was successful

        # Parse the JSON response
        data = response.json()

        # Prepare the CSV file for the current slug
        csv_filename = f"{slug}_Tokens_Data.csv"
        with open(csv_filename, mode='w', newline='', encoding='utf-8') as csv_file:
            writer = csv.writer(csv_file)
            # Write header row
            writer.writerow(["Date", "OP Balance"])

            # Extract and write "OP" token data
            for item in data.get("tokens", []):
                if "date" in item and "tokens" in item:
                    formatted_date = datetime.utcfromtimestamp(item["date"]).strftime('%Y-%m-%d')
                    op_balance = item["tokens"].get("OP", None)  # Get "OP" token balance if it exists
                    if op_balance is not None:
                        writer.writerow([formatted_date, op_balance])

        print(f"OP Tokens Data for '{slug}' saved to {csv_filename}")

    except requests.exceptions.RequestException as e:
        print(f"Error fetching data for {slug}: {e}")
    except Exception as e:
        print(f"An error occurred for {slug}: {e}")
```
