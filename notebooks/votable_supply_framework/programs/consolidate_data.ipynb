```python
import os
import pandas as pd

# Get the current working directory (folder containing this script)
input_folder = os.getcwd()
# Output file path
output_file = os.path.join(input_folder, "../../../data/votable_supply_framework/programs/consolidated_latest_data.csv")

# List to store the data from all CSV files
latest_data = []

# Iterate through all CSV files in the folder
for file_name in os.listdir(input_folder):
    if file_name.endswith("_Tokens_Data.csv"):
        # Extract protocol name from the file name
        protocol_name = file_name.split("_Tokens_Data.csv")[0]

        # Read the CSV file
        file_path = os.path.join(input_folder, file_name)
        df = pd.read_csv(file_path)

        # Skip if the DataFrame is empty or doesn't have required columns
        if df.empty or "Date" not in df.columns or "OP Balance" not in df.columns:
            print(f"Skipping {file_name}: Empty or invalid file.")
            continue

        # Ensure the Date column is in datetime format
        try:
            df["Date"] = pd.to_datetime(df["Date"], errors="coerce")
        except Exception as e:
            print(f"Skipping {file_name}: Error parsing dates - {e}")
            continue

        # Drop rows with invalid dates
        df = df.dropna(subset=["Date"])

        # Skip if the DataFrame is empty after dropping invalid dates
        if df.empty:
            print(f"Skipping {file_name}: No valid dates after cleaning.")
            continue

        # Get the latest date row
        latest_row = df.loc[df["Date"].idxmax()]

        # Append the data to the list
        latest_data.append({
            "Protocol Name": protocol_name,
            "Date": latest_row["Date"],
            "OP Balance": latest_row["OP Balance"]
        })

# Create a DataFrame from the list
output_df = pd.DataFrame(latest_data)

# Save the DataFrame to a CSV file
output_df.to_csv(output_file, index=False)

print(f"Consolidated file saved as {output_file}")
```
