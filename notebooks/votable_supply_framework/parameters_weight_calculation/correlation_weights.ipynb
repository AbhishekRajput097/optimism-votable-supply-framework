```python
import pandas as pd

def calculate_correlation_and_weights(csv_file):
    # Load the dataset
    data = pd.read_csv(csv_file)

    # Ensure the "Date" column is excluded for correlation calculation
    numeric_data = data.select_dtypes(include=['number'])

    # Compute the correlation matrix
    correlation_matrix = numeric_data.corr()

    # Extract correlation with "Votable Supply"
    votable_correlation = correlation_matrix["Votable Supply"]

    # Drop "Votable Supply" itself to only keep parameter correlations
    parameter_correlations = votable_correlation.drop("Votable Supply")

    # Convert to absolute values
    abs_correlations = parameter_correlations.abs()

    # Calculate weights
    total_sum = abs_correlations.sum()
    weights = abs_correlations / total_sum

    # Print results
    print("Correlation with Votable Supply:")
    print(parameter_correlations)
    print("\nWeights for each parameter:")
    print(weights)

    # Return weights for further use
    return weights

if __name__ == "__main__":
    # Example usage
    csv_file = "../../../data/votable_supply_framework/parameters_weight_calculation/parameters_values.csv"  # Replace with your CSV file path
    calculate_correlation_and_weights(csv_file)
```
