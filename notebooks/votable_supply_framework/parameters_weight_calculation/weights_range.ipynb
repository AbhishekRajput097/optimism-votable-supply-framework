```python
import numpy as np

def calculate_weightage_ranges(weights, k=1):
    # Convert weights to a NumPy array
    weight_values = np.array(list(weights.values()))

    # Calculate mean and standard deviation
    mean_weight = np.mean(weight_values)
    std_dev_weight = np.std(weight_values)

    # Calculate ranges for each parameter
    ranges = {
        param: (weight - k * std_dev_weight, weight + k * std_dev_weight)
        for param, weight in weights.items()
    }

    # Print the weight ranges
    print("Weightage Ranges for Each Parameter:")
    for param, (low, high) in ranges.items():
        low = max(low, 0)  # Ensure range does not go below 0
        print(f"{param}: [{low:.4f}, {high:.4f}]")

    return ranges

if __name__ == "__main__":
    # Example usage
    weights = {
        "AVPI": 0.218,
        "PR": 0.135,
        "OP Price": 0.518,
        "LAR": 0.129
    }  # Replace with actual weights from the first script
    calculate_weightage_ranges(weights)
```
