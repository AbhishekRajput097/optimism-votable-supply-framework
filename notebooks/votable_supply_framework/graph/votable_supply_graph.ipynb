```python
import pandas as pd
import plotly.graph_objects as go

data = pd.read_csv('../../../data/votable_supply_framework/graph/votable_supply_data.csv')


fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=data['Date'],
        y=data['Circulating Supply'],
        name='Circulating Supply',
        mode='lines',
        line=dict(color='blue'),
        hoverinfo='x+y',
    )
)


fig.add_trace(
    go.Scatter(
        x=data['Date'],
        y=data['Votable Supply %'],
        name='Votable Supply %',
        mode='lines',
        line=dict(color='green'),
        yaxis='y2',
        hoverinfo='x+y',
    )
)


fig.update_layout(
    title='Circulating Supply and Votable Supply % Over Time',
    xaxis=dict(title='Date'),
    yaxis=dict(
        title='Circulating Supply',
        titlefont=dict(color='blue'),
        tickfont=dict(color='blue'),
    ),
    yaxis2=dict(
        title='Votable Supply %',
        titlefont=dict(color='green'),
        tickfont=dict(color='green'),
        overlaying='y',
        side='right',
    ),
    legend=dict(orientation='h', x=0.5, xanchor='center', y=-0.2),
    hovermode='x',
)


fig.write_html('interactive_plot.html')

fig.show()
```
